```python
import random
import math

# ---------------------------------------
# PARAMETERS
# ---------------------------------------

AREA_SIZE = 100          # 100 x 100 network area
NUM_ROUTERS = 5          # Number of routers to place
NUM_USERS = 50           # Number of users
POPULATION_SIZE = 30
GENERATIONS = 50
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.05
COVERAGE_RADIUS = 25

# Generate random user locations
users = [
    (random.uniform(0, AREA_SIZE), random.uniform(0, AREA_SIZE))
    for _ in range(NUM_USERS)
]


# ---------------------------------------
# DISTANCE FUNCTION
# ---------------------------------------

def distance(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2
    )


# ---------------------------------------
# FITNESS FUNCTION
# ---------------------------------------

def fitness(chromosome):
    """
    Fitness = percentage of users covered
    """

    covered_users = 0

    for user in users:
        for router in chromosome:
            if distance(user, router) <= COVERAGE_RADIUS:
                covered_users += 1
                break

    coverage = covered_users / NUM_USERS

    # Fitness value
    return coverage


# ---------------------------------------
# CREATE CHROMOSOME
# ---------------------------------------

def create_chromosome():

    return [
        (
            random.uniform(0, AREA_SIZE),
            random.uniform(0, AREA_SIZE)
        )
        for _ in range(NUM_ROUTERS)
    ]


# ---------------------------------------
# INITIAL POPULATION
# ---------------------------------------

population = [
    create_chromosome()
    for _ in range(POPULATION_SIZE)
]


# ---------------------------------------
# SELECTION
# ---------------------------------------

def selection(population):

    # Tournament selection
    tournament = random.sample(population, 3)

    return max(
        tournament,
        key=fitness
    )


# ---------------------------------------
# CROSSOVER
# ---------------------------------------

def crossover(parent1, parent2):

    if random.random() > CROSSOVER_RATE:
        return parent1.copy(), parent2.copy()

    point = random.randint(1, NUM_ROUTERS - 1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# ---------------------------------------
# MUTATION
# ---------------------------------------

def mutation(chromosome):

    new_chromosome = chromosome.copy()

    for i in range(NUM_ROUTERS):

        if random.random() < MUTATION_RATE:

            x = random.uniform(0, AREA_SIZE)
            y = random.uniform(0, AREA_SIZE)

            new_chromosome[i] = (x, y)

    return new_chromosome


# ---------------------------------------
# GENETIC ALGORITHM
# ---------------------------------------

best_solution = None
best_fitness = 0

for generation in range(GENERATIONS):

    new_population = []

    while len(new_population) < POPULATION_SIZE:

        # Selection
        parent1 = selection(population)
        parent2 = selection(population)

        # Crossover
        child1, child2 = crossover(parent1, parent2)

        # Mutation
        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)

    population = new_population

    # Find best solution
    generation_best = max(
        population,
        key=fitness
    )

    generation_fitness = fitness(generation_best)

    if generation_fitness > best_fitness:

        best_fitness = generation_fitness
        best_solution = generation_best

    print(
        f"Generation {generation + 1}: "
        f"Best Fitness = {generation_fitness:.4f}"
    )


# ---------------------------------------
# FINAL RESULT
# ---------------------------------------

print("\n========== FINAL SOLUTION ==========")

print(
    f"Best Fitness Value: {best_fitness:.4f}"
)

print(
    f"User Coverage: {best_fitness * 100:.2f}%"
)

print("\nOptimal Router Locations:")

for i, router in enumerate(best_solution):

    print(
        f"Router {i + 1}: "
        f"X = {router[0]:.2f}, "
        f"Y = {router[1]:.2f}"
    )
```
